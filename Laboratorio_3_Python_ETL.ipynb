## Parte 1
### Estructuras Combinadas y Funciones Básicas

```python
# Lista de diccionarios que simula la respuesta de la API de pagos
respuesta_api = [
    {"trx_id": "T001", "monto_usd": 150.00, "estado": "Completado"},
    {"trx_id": "T002", "monto_usd": "Error_Red", "estado": "Fallido"},
    {"trx_id": "T003", "monto_usd": 320.50, "estado": "Completado"},
    {"trx_id": "T004", "monto_usd": 45.00, "estado": "Completado"}
]
```

```python
def extraer_montos(lista_registros):
    # Lista vacía donde se acumularán los montos extraídos
    lista_montos = []
    # Recorremos cada diccionario de la lista de entrada
    for registro in lista_registros:
        # Extraemos el valor de la llave "monto_usd" y lo agregamos a la lista
        lista_montos.append(registro["monto_usd"])
    return lista_montos

# Probamos la función pasándole respuesta_api
montos_extraidos = extraer_montos(respuesta_api)
print("Montos extraídos:", montos_extraidos)
```

## Parte 2
### Manejo de Errores (Red de Seguridad)

```python
def convertir_a_soles(monto):
    # El try intenta la conversión normal; si el monto no es numérico (ej. "Error_Red"),
    # la multiplicación lanzará una excepción que capturamos en el except
    try:
        monto_en_soles = round(monto * 3.8, 2)
        return monto_en_soles
    except Exception as e:
        # Capturamos la excepción general para no detener la ejecución del programa
        print(f"⚠️ Alerta: Dato corrupto detectado. Detalle: {e}")
        # Retornamos 0 para que el resto del flujo pueda continuar sin fallar
        return 0

montos_en_soles = [convertir_a_soles(monto) for monto in montos_extraidos]
print("Montos convertidos a soles:", montos_en_soles)
```

## Parte 3
### List Comprehension (Procesamiento Ágil)

```python
# Lista ya limpia y validada en soles, enviada por el área de Finanzas
montos_limpios = [570.0, 0, 1217.9, 171.0, 850.0, 90.0]

# Calculamos el 18% de impuesto SOLO para los montos mayores a 500.0, en una sola línea
impuestos_grandes = [round(monto * 0.18, 2) for monto in montos_limpios if monto > 500.0]

print("Impuestos de montos grandes:", impuestos_grandes)
print("Cantidad de valores:", len(impuestos_grandes))  # debe confirmar que son 3
```

## Parte 4
### Buenas Prácticas y Reporte Final

```python
# Sumamos el total de impuestos calculados en la Parte 3
suma_impuestos = sum(impuestos_grandes)

# Reporte final usando \n para dar formato de varias líneas en un solo print
print(f"\n--- REPORTE FINAL DE AUDITORÍA ---\nTransacciones grandes procesadas: {len(impuestos_grandes)}\nTotal de impuestos recaudados: S/. {suma_impuestos}\n")
```
